```python

import pandas as pd

summary = pd.read_excel(
    "csec_attendance.xlsx",
    sheet_name="Student Summary"
)

log = pd.read_excel(
    "csec_attendance.xlsx",
    sheet_name="Daily Log"
)

summary["Attendance %"] = pd.to_numeric(
    summary["Attendance %"]
    .astype(str)
    .str.replace("%", "", regex=False),
    errors="coerce"
)

print("\nAll Student Records:")
print(summary.to_string(index=False))

print("\nDataset Information:")
summary.info()

print("\nMissing Values:")
print(summary.isnull().sum())

print("\nStudent Attendance Details:")
print(summary[
    ["Roll No", "Student Name", "Present", "Absent", "Attendance %"]
].to_string(index=False))

print("\nAverage Attendance:")
print(summary["Attendance %"].mean())

print("\nStudents with Attendance Below 75%:")
print(
    summary[summary["Attendance %"] < 75][
        ["Roll No", "Student Name", "Attendance %"]
    ].to_string(index=False)
)

print("\nStudents with Attendance Above or Equal to 75%:")
print(
    summary[summary["Attendance %"] >= 75][
        ["Roll No", "Student Name", "Attendance %"]
    ].to_string(index=False)
)

print("\nAttendance Status Count:")
print(summary["Status"].value_counts())

print("\nAttendance by Subject:")
print(summary.groupby("Subject")["Attendance %"].mean())

print("\nDaily Attendance Count:")
print(log["Status"].value_counts())

print("\nAll Attendance Records:")
print(log.to_string(index=False))
```
